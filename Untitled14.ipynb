```python
from google.colab.output import eval_js
from IPython.display import display, Javascript
import base64
import numpy as np
import cv2
import matplotlib.pyplot as plt

def start_camera():
    js = """
    async function startCamera() {
        const stream = await navigator.mediaDevices.getUserMedia({video: true});
        const video = document.createElement('video');

        video.srcObject = stream;
        video.setAttribute('playsinline', '');
        video.style.width = '400px';

        document.body.appendChild(video);
        await video.play();

        window.colabVideo = video;
        window.colabStream = stream;

        return true;
    }
    """
    display(Javascript(js))
    return eval_js("startCamera()")


def capture_frame():
    js = """
    async function captureFrame() {
        const video = window.colabVideo;

        if (!video || video.videoWidth === 0) {
            throw new Error("Camera is not ready");
        }

        const canvas = document.createElement('canvas');
        canvas.width = video.videoWidth;
        canvas.height = video.videoHeight;

        canvas.getContext('2d').drawImage(video, 0, 0);

        return canvas.toDataURL('image/jpeg');
    }
    """
    display(Javascript(js))

    data = eval_js("captureFrame()")
    image_bytes = base64.b64decode(data.split(',')[1])

    return cv2.imdecode(
        np.frombuffer(image_bytes, np.uint8),
        cv2.IMREAD_COLOR
    )

print("Camera functions defined successfully!")
```

```python
from IPython.display import Javascript, display
from google.colab.output import eval_js
import base64
import numpy as np
import cv2
import matplotlib.pyplot as plt

def start_camera():
    js = """
    window.startCamera = async function() {

        if (window.colabStream) {
            window.colabStream.getTracks().forEach(t => t.stop());
        }

        const stream = await navigator.mediaDevices.getUserMedia({
            video: true
        });

        const video = document.createElement('video');

        video.autoplay = true;
        video.muted = true;
        video.playsInline = true;
        video.style.width = '400px';
        video.srcObject = stream;

        document.body.appendChild(video);

        window.colabStream = stream;
        window.colabVideo = video;

        await new Promise((resolve, reject) => {
            if (video.readyState >= 1) {
                resolve();
            } else {
                video.addEventListener('loadedmetadata', resolve, {once:true});
                setTimeout(() => reject(new Error("Video metadata timeout")), 10000);
            }
        });

        await video.play();

        await new Promise((resolve, reject) => {
            const check = () => {
                if (video.videoWidth > 0 && video.videoHeight > 0) {
                    resolve();
                } else {
                    setTimeout(check, 200);
                }
            };
            check();
            setTimeout(() => reject(new Error("Camera dimensions unavailable")), 10000);
        });

        return video.videoWidth + " x " + video.videoHeight;
    };
    """

    display(Javascript(js))
    result = eval_js("window.startCamera()")
    print("Camera ready:", result)
```

```python
import cv2
from pathlib import Path

RAW_IMAGES = Path("/content/drowsiness_project/data/raw/images")
RAW_LABELS = Path("/content/drowsiness_project/data/raw/labels")

RAW_LABELS.mkdir(parents=True, exist_ok=True)

class_mapping = {
    "awake": 0,
    "drowsy": 1
}

annotated = 0
not_detected = []

for image_path in RAW_IMAGES.glob("*.jpg"):

    filename = image_path.stem.lower()

    # Determine class from filename
    if filename.startswith("awake"):
        class_id = 0
    elif filename.startswith("drowsy"):
        class_id = 1
    else:
        continue

    image = cv2.imread(str(image_path))

    if image is None:
        print("Could not read:", image_path.name)
        continue

    height, width = image.shape[:2]

    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    faces = face_detector.detectMultiScale(
        gray,
        scaleFactor=1.1,
        minNeighbors=5,
        minSize=(40, 40)
    )

    if len(faces) == 0:
        not_detected.append(image_path.name)
        continue

    # Select the largest detected face
    x, y, w, h = max(faces, key=lambda box: box[2] * box[3])

    # Convert bounding box to normalized YOLO coordinates
    x_center = (x + w / 2) / width
    y_center = (y + h / 2) / height
    box_width = w / width
    box_height = h / height

    label_path = RAW_LABELS / f"{image_path.stem}.txt"

    with open(label_path, "w") as file:
        file.write(
            f"{class_id} {x_center:.6f} "
            f"{y_center:.6f} {box_width:.6f} {box_height:.6f}\n"
        )

    annotated += 1

print("Annotations generated:", annotated)
print("Faces not detected:", len(not_detected))
print("Images without detected faces:", not_detected)
```

```python
import cv2
import matplotlib.pyplot as plt
from pathlib import Path
import math

image_files = sorted(
    list(RAW_IMAGES.glob("awake_*.jpg")) +
    list(RAW_IMAGES.glob("drowsy_*.jpg"))
)

cols = 4
rows = math.ceil(len(image_files) / cols)

fig, axes = plt.subplots(rows, cols, figsize=(16, 4 * rows))
axes = axes.flatten() if hasattr(axes, "flatten") else [axes]

for ax, image_path in zip(axes, image_files):

    image = cv2.imread(str(image_path))
    label_path = RAW_LABELS / f"{image_path.stem}.txt"

    if image is None:
        ax.set_title("Image unreadable")
        ax.axis("off")
        continue

    h, w = image.shape[:2]

    if label_path.exists():
        with open(label_path, "r") as f:
            for line in f:
                cls, xc, yc, bw, bh = map(float, line.split())

                x1 = int((xc - bw / 2) * w)
                y1 = int((yc - bh / 2) * h)
                x2 = int((xc + bw / 2) * w)
                y2 = int((yc + bh / 2) * h)

                cv2.rectangle(
                    image, (x1, y1), (x2, y2),
                    (0, 255, 0), 2
                )

                label = "Awake" if int(cls) == 0 else "Drowsy"
                cv2.putText(
                    image, label, (x1, max(y1 - 10, 20)),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.7,
                    (0, 255, 0), 2
                )

    ax.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    ax.set_title(image_path.name)
    ax.axis("off")

for ax in axes[len(image_files):]:
    ax.axis("off")

plt.tight_layout()
plt.show()
```

```python
from pathlib import Path

PUBLIC_DATASET = Path("/content/Drowsiness-Detection-2")

print("Dataset exists:", PUBLIC_DATASET.exists())
print("Train images:", (PUBLIC_DATASET / "train/images").exists())
print("Validation images:", (PUBLIC_DATASET / "valid/images").exists())
print("Test images:", (PUBLIC_DATASET / "test/images").exists())
print("YAML exists:", (PUBLIC_DATASET / "data.yaml").exists())
```

```python
import yaml

PUBLIC_YAML = PUBLIC_DATASET / "data.yaml"

with open(PUBLIC_YAML, "r") as f:
    config = yaml.safe_load(f)

config["path"] = str(PUBLIC_DATASET)
config["train"] = "train/images"
config["val"] = "valid/images"
config["test"] = "test/images"

with open(PUBLIC_YAML, "w") as f:
    yaml.safe_dump(config, f, sort_keys=False)

print("Updated YAML successfully!")
print(config)
```

```python
print("Classes:", config["names"])
print("Number of classes:", config["nc"])
```

```python
import torch
import ultralytics

print("Ultralytics:", ultralytics.__version__)
print("PyTorch:", torch.__version__)
print("CUDA available:", torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
```

```python
from ultralytics import YOLO
from pathlib import Path
import torch

DEVICE = 0 if torch.cuda.is_available() else "cpu"

models_to_train = [
    "yolov8n.pt",
    "yolo11n.pt",
    "yolo26n.pt"
]

for model_name in models_to_train:

    print("\n" + "=" * 60)
    print("Starting training:", model_name)
    print("=" * 60)

    model = YOLO(model_name)

    model.train(
        data=str(PUBLIC_YAML),
        epochs=50,
        imgsz=640,
        batch=8,
        device=DEVICE,
        seed=42,
        deterministic=True,
        patience=10,
        workers=2,
        plots=True,
        project="/content/drowsiness_project/model_comparison",
        name=model_name.replace(".pt", ""),
        exist_ok=True
    )

    print("Completed:", model_name)

print("\nAll model training completed!")
```
